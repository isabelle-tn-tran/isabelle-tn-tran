```python

import pandas as pd

# Simulated research data (not real patient data)
data = {
    "participant": ["P001", "P001", "P001",
                    "P002", "P002", "P002"],
    "timepoint": ["Baseline", "60min", "120min",
                  "Baseline", "60min", "120min"],
    "expected_min": [0, 60, 120, 0, 60, 120],
    "actual_min": [0, 64, 119, 0, 79, None]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Flag records with missing collection timestamps
df["missing_collection_time"] = df["actual_min"].isna()

# Calculate collection timing differences
df["time_difference"] = (
    df["actual_min"] - df["expected_min"]
).abs()

# Flag samples outside a 10-minute window
df["outside_window"] = df["time_difference"] > 10

# Create an overall QC status
df["qc_status"] = "PASS"

df.loc[df["outside_window"], "qc_status"] = "LATE_OR_EARLY"
df.loc[df["missing_collection_time"], "qc_status"] = "MISSING_TIME"

print(df[["participant", "timepoint", "qc_status"]])
```

```python
df["qc_status"].value_counts()
```

```python
print(df.columns)
```

```python
proportion_missing = df["missing_collection_time"].sum() / len(df)
percentage_missing = proportion_missing * 100
rounded_percentage = round(percentage_missing, 2)
print(rounded_percentage)
```
